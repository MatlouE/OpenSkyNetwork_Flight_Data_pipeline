```python
import pandas as pd
import requests
```

```python
url = "https://opensky-network.org/api/states/all"
url
```

```python
def fetch_flight_data(url):
    try:
        print("fetching data")
        response = requests.get(url, timeout=10)
        response.raise_for_status()

        data = response.json()
        return data.get('states', [])

    except Exception as e:
        print(f'Error fetching data: {e}')
        return []
```

```python
raw_list = fetch_flight_data(url)
```

```python
raw_list[0]
```

```python
def clean_data(raw_list):
    #we want to store our data in dict format 
    # list of dictionaries
    cleaned_list = []

    #mapping data columns to values 
    for flight in raw_list:
        record = {
            "icao24": flight[0],
            "callsign": flight[1].strip() if flight[1] else "N/A",
            "origin_country": flight[2],
            "longitude" : flight[5],
            "latitude" : flight[6],
            "velocity": flight[9] if flight[9] is not None else 0.0
        }
        cleaned_list.append(record)
    return cleaned_list
```

```python
cleaned_list = clean_data(raw_list)

cleaned_list[0]
```

```python
for record in cleaned_list:
    print(record)
```

```python
flight_df = pd.DataFrame(cleaned_list, columns=["icao24", "callsign", "origin_country", "longitude", "latitude", "velocity"])
```
